EDGES

```python
df = pd.read_parquet(os.path.join(dir_in, f"poi_is_in_zone_{region_str}.parquet")) 
df.head()
```

```python
uid_columns = [col for col in df.columns if col.startswith('unique_id')]
df_cleaned = df.dropna(subset=uid_columns).copy()
df_cleaned.drop_duplicates(inplace=True)
for col in uid_columns:
    df_cleaned[col] = df_cleaned[col].apply(lambda x: int(x) - 1)
df_cleaned.to_parquet(os.path.join(edges_dir, "poi_is_in_zone.parquet"))
df_cleaned.head()
```
