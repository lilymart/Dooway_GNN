```python
dataset = "dooway"
region = "Basilicata" # "EmiliaRomagna"
region_str = region.lower() # "emilia_romagna"
base_dir = os.path.join(get_base_dir(), region)
subset_dir = os.path.join(base_dir, "subset")
nodes_dir = os.path.join(subset_dir, "heterodata", "features", "dataframes")
edges_dir = os.path.join(subset_dir, "heterodata", "edgelists", "dataframes")
```

```python
df_UR= pd.read_parquet(os.path.join(edges_dir, "user_wrote_review.parquet")) #unique_id_user, unique_id_review
df_UR.head()
```

```python
k=8
df_gt = pd.read_parquet(os.path.join(subset_dir, f"users_avg_multilabel_k{k}.parquet"))
df_gt.head()
```

```python
df_gt.shape
```

```python
df_R_ok = df_R[['unique_id', 'review_text_embedding']]

df_result = df_UR.merge(
    df_R_ok,
    left_on='unique_id_review',
    right_on='unique_id',
    how='left'
)

df_result = df_result.drop(columns=['unique_id'])
df_result = df_result[['unique_id_user', 'review_text_embedding']]
#df_result.to_parquet(os.path.join(subset_dir, "competitors", "user_reviews_embs.parquet"), index=False)
df_result.head()
```

```python
df_user_embeddings = (
    df_result
    .groupby("unique_id_user")["review_text_embedding"]
    .apply(lambda embeddings: np.mean(np.vstack(embeddings), axis=0))
    .reset_index()
)

df_user_embeddings.head()
```

```python
df_user_embeddings.shape
```

```python
df_user_embeddings["unique_id_user"].max()
```

```python
# 1. Compute the global mean embedding across all users that have embeddings
global_mean_embedding = np.mean(
    np.vstack(df_user_embeddings["review_text_embedding"].values),
    axis=0
)

# 2. Merge keeping all rows from df_gt
df_final = df_gt.merge(df_user_embeddings, on="unique_id_user", how="left")

# 3. Fill missing embeddings with the global mean
df_final["review_text_embedding"] = df_final["review_text_embedding"].apply(
    lambda x: global_mean_embedding if isinstance(x, float) and np.isnan(x) else x
)

df_final.to_parquet(os.path.join(subset_dir, "competitors", "user_reviews_embs.parquet"), index=False)
df_final.head()
```
